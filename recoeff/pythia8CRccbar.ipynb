```python
#%%capture
%run ../dca/input/Format.ipynb
import ROOT as root
from array import array
root.gErrorIgnoreLevel = root.kFatal
%jsroot on
root.gStyle.SetOptStat(0)
```

```python
colors=[1,2,4,root.kGreen+2,root.kOrange+4,root.kMagenta,root.kGray,root.kGray,root.kBlack,root.kCyan,root.kOrange,root.kViolet,root.kCyan,root.kGreen,root.kOrange]
file_path="input/"
file_names=["ccbarLambdanewtree2500x10000.root"]#,"m_ee_DCA_535.root"],"m_ee_DCA_536.root"]
hist_names=["counts","w_pair","src_count",\
            "mee_exact2_src","mee_star_src","mee_phenix_eta05_src","mee_phenix_y035_src","mee_phenix_phiacc_src","mee_gate_wpair_src","mee_gate_tree_src"\
            ,"pairpt_gate_src","pt_star_src","pt_phenix_eta_src","pt_phenix_y_src","pt_phenix_phi_src","pTHat_src","ptMother_src","ptMother_BRweight_src"]
N_type = len(hist_names)
```

```python
hists_read = []

for iFile in range(len(file_names)):
    infile = root.TFile.Open(file_path+file_names[iFile], "read")
    hists_emc_read0 = []
    for itype in range(N_type):
        hists_emc_read0.append(infile.Get(hist_names[itype]))
        try: hists_emc_read0[-1].SetDirectory(root.nullptr); hists_emc_read0[-1].SetName(hists_emc_read0[-1].GetName()+f"_{iFile}")
        except: 
            print(f"Error setting directory for {hist_names[itype]}{0} in file {file_names[iFile]}")
            break
        hists_emc_read0[-1].SetName(hists_emc_read0[-1].GetName()+f"{iFile}")
    hists_read.append(hists_emc_read0)

    infile.Close()
```

```python
iFile=0
ihist=-1
c1 = root.TCanvas("c1","c1",1400,800)
hist_pt_charm_hadrons = hists_read[iFile][ihist].Rebin(1,f"hist_pt_charm_hadrons_{iFile}")
hist_pt_charm_hadrons.Draw("colz")

c1.Draw()    
```

```python
iFile=0
ihist=-1
c2 = root.TCanvas("c2","c2",1400,800)
hist_pt_charm_hadrons = hists_read[iFile][ihist].Rebin(1,f"hist_pt_charm_hadrons_{iFile}")
hist_pt_D0 = hist_pt_charm_hadrons.ProjectionX(f"hist_pt_D0_{iFile}",1,1)
hist_pt_Dpm = hist_pt_charm_hadrons.ProjectionX(f"hist_pt_Dpm_{iFile}",2,2)
hist_pt_Ds = hist_pt_charm_hadrons.ProjectionX(f"hist_pt_Ds_{iFile}",3,3)
hist_pt_LambdaC = hist_pt_charm_hadrons.ProjectionX(f"hist_pt_LambdaC_{iFile}",4,4)

root.gPad.SetLogy()
Format_Hist_total(hist_pt_D0, "p_{T} of HF e^{#pm} (GeV/c)", "Counts", Lcolor=colors[0], Lwidth=3, left=0.15, bottom=0.15, right=0.01, top=0.01,  Tsize=0.07,  Lsize=0.06,\
                      Mstyle=21,  Msize=2, Mcolor=4, offset_x=0.9, offset_y=1, title="",  Malpha=1,  Lalpha=1)
Format_Graph(hist_pt_Dpm, Lcolor=colors[1], Lwidth=3)
Format_Graph(hist_pt_Ds, Lcolor=colors[2], Lwidth=3)
Format_Graph(hist_pt_LambdaC, Lcolor=colors[3], Lwidth=3)

hist_pt_D0.Draw("hist")
hist_pt_Dpm.Draw("same hist")
hist_pt_Ds.Draw("same hist")
hist_pt_LambdaC.Draw("same hist")

hist_pt_D0.SetMaximum(hist_pt_Dpm.GetMaximum()*2.2)

TLegend = root.TLegend(0.7,0.4,0.99,0.99)
TLegend.SetBorderSize(0)
TLegend.SetFillStyle(0)
TLegend.AddEntry(hist_pt_D0, "D^{0}", "l")
TLegend.AddEntry(hist_pt_Dpm, "D^{#pm}", "l")
TLegend.AddEntry(hist_pt_Ds, "D_{s}", "l")
TLegend.AddEntry(hist_pt_LambdaC, "#Lambda_{c}", "l")
TLegend.Draw()

c2.Draw()
```

```python
iFile=0
ihist=-1
c3 = root.TCanvas("c3","c3",1400,800)
hist_pt_charm_hadrons = hists_read[iFile][ihist].Rebin(1,f"hist_pt_charm_hadrons_{iFile}_{ihist}")
hist_pt_D0 = hist_pt_charm_hadrons.ProjectionX(f"hist_pt_for_ratio_D0_{iFile}",1,1)
hist_pt_Dpm = hist_pt_charm_hadrons.ProjectionX(f"hist_pt_for_ratio_Dpm_{iFile}",2,2)
hist_pt_Ds = hist_pt_charm_hadrons.ProjectionX(f"hist_pt_for_ratio_Ds_{iFile}",3,3)
hist_pt_LambdaC = hist_pt_charm_hadrons.ProjectionX(f"hist_pt_for_ratio_LambdaC_{iFile}",4,4)

Format_Hist_total(hist_pt_D0, "p_{T} of HF e^{#pm} (GeV/c)", "Ratio to D^{#pm}", Lcolor=colors[0], Lwidth=3, left=0.15, bottom=0.15, right=0.01, top=0.01,  Tsize=0.07,  Lsize=0.06,\
                      Mstyle=21,  Msize=2, Mcolor=4, offset_x=0.9, offset_y=1, title="",  Malpha=1,  Lalpha=1)
Format_Graph(hist_pt_Dpm, Lcolor=colors[1], Lwidth=3)
Format_Graph(hist_pt_Ds, Lcolor=colors[2], Lwidth=3)
Format_Graph(hist_pt_LambdaC, Lcolor=colors[3], Lwidth=3)

hist_pt_D0.Divide(hist_pt_Dpm)
hist_pt_Ds.Divide(hist_pt_Dpm)
hist_pt_LambdaC.Divide(hist_pt_Dpm)

hist_pt_D0.Draw("hist")
hist_pt_Ds.Draw("same hist")
hist_pt_LambdaC.Draw("same hist")

hist_pt_D0.GetYaxis().SetRangeUser(0, 0.59)

TLegend = root.TLegend(0.3,0.37,0.9,0.5)
TLegend.SetNColumns(3)
TLegend.SetBorderSize(0)
TLegend.SetFillStyle(0)
TLegend.AddEntry(hist_pt_D0, "D^{0}", "l")
TLegend.AddEntry(hist_pt_Ds, "D_{s}", "l")
TLegend.AddEntry(hist_pt_LambdaC, "#Lambda_{c}", "l")
TLegend.Draw()

c3.Draw()
```
