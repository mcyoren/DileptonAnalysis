```python
colors=[1,2,4,root.kGreen+2,root.kOrange+4,root.kMagenta,root.kGray,root.kGray,root.kBlack,root.kCyan,root.kOrange,root.kViolet,root.kCyan,root.kGreen,root.kOrange]
file_path="input/"
file_names=["ccbarpythia6_v1.root","ccbarpythia6.root"]#,"m_ee_DCA_535.root"],"m_ee_DCA_536.root"]
ee_hist_names=["hMee_all","hMee_PHENIX","hMee_STAR","hMeeVsProcess_all","hMeeVsProcess_PHENIX","hMeeVsProcess_STAR"]
pt_hist_names=["hPtMother_all","hPtMother_PHENIX","hPtMother_STAR","hPtMother_BRweight_all","hPtMother_BRweight_PHENIX","hPtMother_BRweight_STAR","hPtCharm_all","hPtCharm_BRweight","hPtCharmVsProcess"]
forward_hist_names=["hDielectronDphi_forwardIM_eBR_counts","hDielectronDphi_forwardIM_muBR_counts","hDielectronDphi_forwardIM_eBR_dSigma_dPhi","hDielectronDphi_forwardIM_muBR_dSigma_dPhi","hDielectronDphiVsProcess_forwardIM_eBR_counts","hDielectronDphiVsProcess_forwardIM_muBR_counts","hDielectronDphiVsProcess_forwardIM_eBR_dSigma_dPhi","hDielectronDphiVsProcess_forwardIM_muBR_dSigma_dPhi","hDielectronForwardIMInfo"]
star_d0_hist_names=["hSTARCharmCounts_y1", "hSTARD0Counts_y1", "hSTARDstarCounts_y1", "hSTAR_ccbar_d2sigma_y1", "hSTAR_ccbar_d2sigma_y1_half", "hSTAR_ccbar_from_D0_y1", "hSTAR_ccbar_from_Dstar_y1"]
event_info_hist_names = ["hRunSummaryCounts","hRunSummaryWeights"]
single_pt_hist_names=["hSingleElectronPt_midY_counts","hSingleElectronPt_midY_dSigma_dpT","hSingleElectronPtVsProcess_midY","hSingleElectronPtVsMother_midY","hSingleElectronPtVsMother_midY_BRweight","hSingleElectronInfo_midY"]
hist_names = [ee_hist_names, pt_hist_names, forward_hist_names, star_d0_hist_names, event_info_hist_names, single_pt_hist_names]
```

```python
hists_read = []

for iFile in range(len(file_names)):
    infile = root.TFile.Open(file_path+file_names[iFile], "read")
    hists_read_file = []
    for itype in range(len(hist_names)):
        hist_read_type = []
        for iname in hist_names[itype]: 
            hist_read_type.append(infile.Get(iname))
            try: hist_read_type[-1].SetDirectory(root.nullptr); hist_read_type[-1].SetName(hist_read_type[-1].GetName()+f"_{iFile}")
            except: 
                print(f"Error setting directory for {hist_names[itype]}{0} in file {file_names[iFile]}")
                break

            hist_read_type[-1].SetName(hist_read_type[-1].GetName()+f"{iFile}")

        hists_read_file.append(hist_read_type)
        
    hists_read.append(hists_read_file)

    infile.Close()
```
