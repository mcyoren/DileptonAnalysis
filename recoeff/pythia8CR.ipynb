```python
#%%capture
%run ../dca/input/Format.ipynb
import ROOT as root
from array import array
root.gErrorIgnoreLevel = root.kFatal
%jsroot on
root.gStyle.SetOptStat(0)
```

```python
colors=[1,2,4,root.kGreen+2,root.kOrange+4,root.kMagenta,root.kGray,root.kGray,root.kBlack,root.kCyan,root.kOrange,root.kViolet,root.kCyan,root.kGreen,root.kOrange]
file_path="input/"
file_names=["ccbarLambdanewtree00004.root","ccbarLambdanewtree00005.root"]#,"m_ee_DCA_535.root"],"m_ee_DCA_536.root"]
hist_names=["pt_lambda","pt_lambdac","pt_k0s","pt_d0"]
N_type = len(hist_names)
```

```python
hists_read = []

for iFile in range(len(file_names)):
    infile = root.TFile.Open(file_path+file_names[iFile], "read")
    hists_emc_read0 = []
    for itype in range(N_type):
        hists_emc_read0.append(infile.Get(hist_names[itype]))
        try: hists_emc_read0[-1].SetDirectory(root.nullptr); hists_emc_read0[-1].SetName(hists_emc_read0[-1].GetName()+f"_{iFile}")
        except: 
            print(f"Error setting directory for {hist_names[itype]}{0} in file {file_names[iFile]}")
            break
        hists_emc_read0[-1].SetName(hists_emc_read0[-1].GetName()+f"{iFile}")
    hists_read.append(hists_emc_read0)

    infile.Close()
```

```python
iFile=1
c1 = root.TCanvas("c1","c1",1400,800)
hist_pt_Lambda_to_k0s = hists_read[iFile][0].Rebin(1,f"hist_pt_Lambda_to_k0s_{iFile}")
hist_pt_Lambda_to_k0s.Divide(hists_read[iFile][2])
hist_pt_Lambda_to_k0s.SetLineColor(colors[iFile])
hist_pt_Lambda_to_k0s.SetLineWidth(4)
hist_pt_Lambda_to_k0s.SetTitle("Pythia8 Detroit tune")
hist_pt_Lambda_to_k0s.GetXaxis().SetTitle("p_{T} (GeV/c)")
hist_pt_Lambda_to_k0s.GetYaxis().SetTitle("Lambda/K0s")
hist_pt_Lambda_to_k0s.GetXaxis().SetRangeUser(0,5)
hist_pt_Lambda_to_k0s.Scale(0.5)
hist_pt_Lambda_to_k0s.Draw("HIST")
hist_pt_Lambdac_to_d0 = hists_read[iFile][1].Rebin(1,f"hist_pt_Lambdac_to_d0_{iFile}")
hist_pt_Lambdac_to_d0.Divide(hists_read[iFile][3])
hist_pt_Lambdac_to_d0.SetLineColor(colors[iFile+1])
hist_pt_Lambdac_to_d0.SetLineWidth(4)
hist_pt_Lambdac_to_d0.Draw("SAME HIST")

legend = root.TLegend(0.6,0.7,0.9,0.9)
legend.AddEntry(hist_pt_Lambda_to_k0s, "#Lambda/K^{0}_{s}", "l")
legend.AddEntry(hist_pt_Lambdac_to_d0, "#Lambda_{c}/D^{0}", "l")
legend.Draw()


c1.Draw()    
```

```python
iFile=0
c1 = root.TCanvas("c1","c1",1400,800)
CR_hist_pt_Lambda_to_k0s = hists_read[iFile][0].Rebin(1,f"CR_hist_pt_Lambda_to_k0s_{iFile}")
CR_hist_pt_Lambda_to_k0s.Divide(hists_read[iFile][2])
CR_hist_pt_Lambda_to_k0s.SetLineColor(colors[iFile])
CR_hist_pt_Lambda_to_k0s.SetLineWidth(4)
CR_hist_pt_Lambda_to_k0s.SetTitle("Pythia8 CR")
CR_hist_pt_Lambda_to_k0s.GetXaxis().SetTitle("p_{T} (GeV/c)")
CR_hist_pt_Lambda_to_k0s.GetYaxis().SetTitle("Lambda/K0s")
CR_hist_pt_Lambda_to_k0s.GetXaxis().SetRangeUser(0,5)
CR_hist_pt_Lambda_to_k0s.Scale(0.5)
CR_hist_pt_Lambda_to_k0s.Draw("HIST")
CR_hist_pt_Lambdac_to_d0 = hists_read[iFile][1].Rebin(1,f"CR_hist_pt_Lambdac_to_d0_{iFile}")
CR_hist_pt_Lambdac_to_d0.Divide(hists_read[iFile][3])
CR_hist_pt_Lambdac_to_d0.SetLineColor(colors[iFile+1])
CR_hist_pt_Lambdac_to_d0.SetLineWidth(4)
CR_hist_pt_Lambdac_to_d0.Draw("SAME HIST")

legend = root.TLegend(0.6,0.7,0.9,0.9)
legend.AddEntry(CR_hist_pt_Lambda_to_k0s, "#Lambda/K^{0}_{s}", "l")
legend.AddEntry(CR_hist_pt_Lambdac_to_d0, "#Lambda_{c}/D^{0}", "l")
legend.Draw()


c1.Draw()    
```

```python
from array import array

def make_graph(name, pt, ratio):
    x = array("d", pt)
    y = array("d", ratio)
    g = root.TGraph(len(pt), x, y)
    g.SetName(name)
    g.SetTitle(";p_{T} [GeV/c];#Lambda/K^{0}_{S}")
    return g

g_black = make_graph("g_STAR_pp200_LambdaK0S_dNch_1p97", pt_black, lam_k0s_black)
g_blue  = make_graph("g_STAR_pp200_LambdaK0S_dNch_4p68", pt_blue,  lam_k0s_blue)
g_red   = make_graph("g_STAR_pp200_LambdaK0S_dNch_9p01", pt_red,   lam_k0s_red)
```

```python
c3 = root.TCanvas("c3", "c3", 1400, 800)
g_black.SetMarkerStyle(20)
g_black.SetMarkerColor(root.kBlack)
g_black.SetMarkerSize(2)
g_black.SetLineColor(root.kBlack)
g_black.SetLineWidth(2)
g_black.Draw("AP")
g_black.GetYaxis().SetRangeUser(0,1.2)
g_blue.SetMarkerStyle(21)
g_blue.SetMarkerColor(root.kBlue)
g_blue.SetLineColor(root.kBlue)
g_blue.SetLineWidth(2)
g_blue.SetMarkerSize(2)
g_blue.Draw("P SAME")
g_red.SetMarkerStyle(22)
g_red.SetMarkerColor(root.kRed)
g_red.SetLineColor(root.kRed)
g_red.SetLineWidth(2)
g_red.SetMarkerSize(2)
g_red.Draw("P SAME") 


hist_pt_Lambda_to_k0s.Draw("SAME HIST")
CR_hist_pt_Lambda_to_k0s.Draw("SAME HIST")

legend = root.TLegend(0.1,0.7,0.4,0.9)
legend.AddEntry(g_black, "STAR dN_{ch}/d#eta=1.97", "P")
legend.AddEntry(g_blue, "STAR dN_{ch}/d#eta=4.68", "P")
legend.AddEntry(g_red, "STAR dN_{ch}/d#eta=9.01", "P")
legend.AddEntry(hist_pt_Lambda_to_k0s, "Pythia8 Detroit tune", "l")
legend.AddEntry(CR_hist_pt_Lambda_to_k0s, "Pythia8 CR", "l")
legend.Draw()

c3.Draw()
```
