```python
def root_array(values):
    """Convert a Python or NumPy array into array('d')."""
    return array("d", np.asarray(values, dtype=float))


def make_asymmetric_graph(
    x,
    y,
    xerr_low,
    xerr_high,
    color,
    line_width=2,
    marker=False,
    marker_size=1.15,
):
    """
    Create a TGraphAsymmErrors with horizontal asymmetric errors.
    """

    x = np.asarray(x, dtype=float)
    y = np.asarray(y, dtype=float)

    n = len(x)

    zero = np.zeros(n, dtype=float)

    graph = root.TGraphAsymmErrors(
        n,
        root_array(x),
        root_array(y),
        root_array(xerr_low),
        root_array(xerr_high),
        root_array(zero),
        root_array(zero),
    )

    graph.SetLineColor(color)
    graph.SetLineWidth(line_width)

    if marker:
        graph.SetMarkerStyle(20)
        graph.SetMarkerSize(marker_size)
        graph.SetMarkerColor(root.kBlack)
    else:
        graph.SetMarkerStyle(1)
        graph.SetMarkerSize(0)

    return graph

def draw_multiline_label(
    x,
    y,
    text,
    main_text_size=0.030,
    reference_text_size=0.023,
    line_spacing=0.36,
    font=42,
):
    """
    Draw labels split at the | character.

    The first line is treated as the main measurement label.
    The second line is treated as the smaller reference.

    Example:
        "PHENIX, p+p|PRL 96, 032001"
    """

    latex_objects = []
    lines = text.split("|")

    if len(lines) == 1:
        latex = root.TLatex()
        latex.SetTextFont(font)
        latex.SetTextSize(main_text_size)
        latex.SetTextAlign(12)
        latex.DrawLatex(x, y, lines[0])
        latex_objects.append(latex)

        return latex_objects

    # Main label, slightly above the row center
    main_label = root.TLatex()
    main_label.SetTextFont(font)
    main_label.SetTextSize(main_text_size)
    main_label.SetTextAlign(12)
    main_label.DrawLatex(
        x,
        y + 0.5 * line_spacing,
        lines[0],
    )
    latex_objects.append(main_label)

    # Reference, smaller and slightly below
    reference_label = root.TLatex()
    reference_label.SetTextFont(font)
    reference_label.SetTextSize(reference_text_size)
    reference_label.SetTextAlign(12)
    reference_label.DrawLatex(
        x,
        y - 0.5 * line_spacing,
        lines[1],
    )
    latex_objects.append(reference_label)

    return latex_objects


def draw_vertical_line(
    x,
    ymin,
    ymax,
    color,
    style=1,
    width=2,
):
    line = root.TLine(x, ymin, x, ymax)
    line.SetLineColor(color)
    line.SetLineStyle(style)
    line.SetLineWidth(width)
    line.Draw()
    return line


def draw_cross_section_plot(
    canvas_name,
    values,
    stat_low,
    stat_high,
    total_low,
    total_high,
    labels,
    xmin,
    xmax,
    xlabel,
    title,
    label_x,
    separator_after=None,
    canvas_width=850,
    canvas_height=850,
):
    """
    Create the base ROOT cross-section plot.

    Returns a dictionary containing the canvas and ROOT objects.
    Keeping this dictionary prevents ROOT objects from being
    garbage-collected in a notebook.
    """

    values = np.asarray(values, dtype=float)

    n = len(values)

    # Top row receives the largest y coordinate.
    y = np.arange(n, 0, -1, dtype=float)

    ymin = 0.0
    ymax = n + 1.0

    canvas = root.TCanvas(
        canvas_name,
        canvas_name,
        canvas_width,
        canvas_height,
    )

    canvas.SetLeftMargin(0.08)
    canvas.SetRightMargin(0.02)
    canvas.SetBottomMargin(0.11)
    canvas.SetTopMargin(0.11)

    frame = canvas.DrawFrame(xmin, ymin, xmax, ymax)

    frame.GetXaxis().SetTitle(xlabel)
    frame.GetXaxis().CenterTitle(False)

    frame.GetXaxis().SetTitleSize(0.050)
    frame.GetXaxis().SetTitleOffset(1.00)
    frame.GetXaxis().SetLabelSize(0.038)

    # Draw y-axis ticks but suppress numerical labels.
    frame.GetYaxis().SetNdivisions(n, False)
    frame.GetYaxis().SetLabelSize(0.0)
    frame.GetYaxis().SetTickLength(0.018)

    # Total uncertainty first, so it remains behind the red bars.
    graph_total = make_asymmetric_graph(
        values,
        y,
        total_low,
        total_high,
        root.kBlue,
        line_width=2,
        marker=False,
    )

    graph_total.Draw("[] SAME")

    # Statistical uncertainty.
    graph_stat = make_asymmetric_graph(
        values,
        y,
        stat_low,
        stat_high,
        root.kRed + 1,
        line_width=2,
        marker=False,
    )

    graph_stat.Draw("[] SAME")

    # Central points.
    graph_points = root.TGraph(
        n,
        root_array(values),
        root_array(y),
    )

    graph_points.SetMarkerStyle(20)
    graph_points.SetMarkerSize(1.15)
    graph_points.SetMarkerColor(root.kBlack)
    graph_points.SetLineColor(root.kBlack)
    graph_points.Draw("P SAME")

    objects = {
        "canvas": canvas,
        "frame": frame,
        "graph_total": graph_total,
        "graph_stat": graph_stat,
        "graph_points": graph_points,
        "labels": [],
        "lines": [],
        "boxes": [],
        "legends": [],
    }

    # Measurement labels.
    for yi, label in zip(y, labels):
        label_objects = draw_multiline_label(
            label_x,
            yi,
            label,
            text_size=0.026,
            line_spacing=0.35,
        )

        objects["labels"].extend(label_objects)

    # Horizontal dashed separator.
    if separator_after is not None:
        separator_y = y[separator_after - 1] - 0.62

        separator = root.TLine(
            xmin,
            separator_y,
            xmax,
            separator_y,
        )

        separator.SetLineColor(root.kGray + 2)
        separator.SetLineStyle(2)
        separator.SetLineWidth(1)
        separator.Draw()

        objects["lines"].append(separator)

    # Plot title.
    title_latex = root.TLatex()
    title_latex.SetNDC(True)
    title_latex.SetTextFont(62)
    title_latex.SetTextSize(0.031)
    title_latex.SetTextAlign(13)
    title_latex.DrawLatex(0.095, 0.955, title)

    objects["labels"].append(title_latex)

    canvas.RedrawAxis()
    canvas.Modified()
    canvas.Update()

    return objects
```

```python
def draw_cross_section_plot(
    canvas_name,
    values,
    stat_low,
    stat_high,
    total_low,
    total_high,
    labels,
    xmin,
    xmax,
    xlabel,
    title,
    label_x,
    old_results_count,
    draw_new_pp=True,
    canvas_width=850,
    canvas_height=850,
):
    """
    Draw cross-section measurements.

    Parameters
    ----------
    old_results_count : int
        Number of old measurements above the dotted line.

    draw_new_pp : bool
        True:
            Draw old measurements and new p+p results, including the
            dotted separator.

        False:
            Draw only old measurements above the dotted line.
    """

    values = np.asarray(values, dtype=float)
    stat_low = np.asarray(stat_low, dtype=float)
    stat_high = np.asarray(stat_high, dtype=float)
    total_low = np.asarray(total_low, dtype=float)
    total_high = np.asarray(total_high, dtype=float)

    # Keep only measurements above the dotted line.
    if not draw_new_pp:
        values = values[:old_results_count]
        stat_low = stat_low[:old_results_count]
        stat_high = stat_high[:old_results_count]
        total_low = total_low[:old_results_count]
        total_high = total_high[:old_results_count]
        labels = labels[:old_results_count]

    n = len(values)

    # First entry is drawn at the top.
    y = np.arange(n, 0, -1, dtype=float)

    ymin = 0.0
    ymax = n + 1.5

    canvas = root.TCanvas(
        canvas_name,
        canvas_name,
        canvas_width,
        canvas_height,
    )

    canvas.SetLeftMargin(0.08)
    canvas.SetRightMargin(0.02)
    canvas.SetBottomMargin(0.11)
    canvas.SetTopMargin(0.11)

    frame = canvas.DrawFrame(xmin, ymin, xmax, ymax)

    frame.GetXaxis().SetTitle(xlabel)
    frame.GetXaxis().SetTitleSize(0.050)
    frame.GetXaxis().SetTitleOffset(1.00)
    frame.GetXaxis().SetLabelSize(0.038)

    frame.GetYaxis().SetNdivisions(n, False)
    frame.GetYaxis().SetLabelSize(0.0)
    frame.GetYaxis().SetTickLength(0.018)

    # Blue total uncertainty.
    graph_total = make_asymmetric_graph(
        values,
        y,
        total_low,
        total_high,
        root.kBlue,
        line_width=2,
        marker=False,
    )
    graph_total.Draw("[] SAME")

    # Red statistical uncertainty.
    graph_stat = make_asymmetric_graph(
        values,
        y,
        stat_low,
        stat_high,
        root.kRed + 1,
        line_width=2,
        marker=False,
    )
    graph_stat.Draw("[] SAME")

    # Central values.
    graph_points = root.TGraph(
        n,
        root_array(values),
        root_array(y),
    )

    graph_points.SetMarkerStyle(20)
    graph_points.SetMarkerSize(1.15)
    graph_points.SetMarkerColor(root.kBlack)
    graph_points.SetLineColor(root.kBlack)
    graph_points.Draw("P SAME")

    objects = {
        "canvas": canvas,
        "frame": frame,
        "graph_total": graph_total,
        "graph_stat": graph_stat,
        "graph_points": graph_points,
        "labels": [],
        "lines": [],
        "boxes": [],
        "legends": [],
        "npoints": n,
        "ymin": ymin,
        "ymax": ymax,
        "draw_new_pp": draw_new_pp,
    }

    for yi, label in zip(y, labels):
        label_objects = draw_multiline_label(
            label_x,
            yi,
            label,
            main_text_size=0.040,
            reference_text_size=0.025,
            line_spacing=0.36,
            font=42,
        )
        objects["labels"].extend(label_objects)

    # Draw separator only when both old and new results are shown.
    if draw_new_pp:
        separator_y = y[old_results_count - 1] - 0.62

        separator = root.TLine(
            xmin,
            separator_y,
            xmax,
            separator_y,
        )

        separator.SetLineColor(root.kGray + 2)
        separator.SetLineStyle(2)
        separator.SetLineWidth(1)
        separator.Draw()

        objects["lines"].append(separator)

    title_latex = root.TLatex()
    title_latex.SetNDC(True)
    title_latex.SetTextFont(62)
    title_latex.SetTextSize(0.045)
    title_latex.SetTextAlign(13)
    title_latex.DrawLatex(0.095, 0.88, title)

    objects["labels"].append(title_latex)

    canvas.RedrawAxis()
    canvas.Modified()
    canvas.Update()

    return objects
```
